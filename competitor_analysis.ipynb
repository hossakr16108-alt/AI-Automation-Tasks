```python
websites = [
    "https://www.coursera.org/",
    "https://www.edx.org/",
    "https://www.udemy.com/"
]
```

```python
website_contents = []

for url in websites:
    content = fetch_website_contents(url)
    website_contents.append(content)
```

```python
system_prompt = """
You are a website analysis assistant.

Analyze the provided website content and identify:
1. Strengths of the website
2. Weaknesses of the website

Focus on useful aspects such as content quality, organization, usability,
clarity, design and information.

Respond in clear Markdown.
Do not wrap the response in a code block.
"""
```

```python
user_prompt = """
Analyze this website and give me its strengths and weaknesses.

Website URL:
{url}

Website content:
{website}
"""
```

```python
def analyze_website(url):
    website = fetch_website_contents(url)

    prompt = user_prompt.format(
        url=url,
        website=website
    )

    response = ollama.chat.completions.create(
        model="gemma3:270m",
        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content
```

```python
for url in websites:
    print(f"\n{url}\n")
    analysis = analyze_website(url)
    display(Markdown(analysis))
```
